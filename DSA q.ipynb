```python
TC = TIME COMPLEXITY
SC = SPACE COMPLEXITY
```

```python
N= int(input("Enter the number you want:- "))
count=0
while N>0:                                                 #TC = O(Log base 10 n) "As the number is devided by 10"
    N= N//10                                               #SC = O(1)
    count+=1                              
print(f"The count of that numbers digit is", count)
```

```python
N=int(input("Enter the number you want to check:-"))
num=N
rev=0
while num>0:
    ld = num % 10                                       #TC = O(Log base 10 n) "As the number is devided by 10"
    rev = (rev * 10) + ld                               #SC = O(1)
    num //= 10
if N == rev:
    print("The number you entered is a Palindrome")
else:
    print("The number you entered is not a Palindrome")
```

```python
N = int(input("Enter the number you want to Check:- "))
num = N
nod = len(str(N))
total = 0
while num > 0:                                                 #TC = O(Log base 10 n) "As the number is devided by 10"
    ld = num % 10                                              #SC = O(1)
    total = total + (ld**nod)                                  
    num = num // 10
if N == total:
    print("The number you entered is an ARMSTRONG number")
else :
    print("The number you entred is not an ARMSTRONG number")
```

```python
def fun(n):
    if n == 0:
        return             #Head recursion
                    
                               # T.c = O(n)      
                               # s.c = O(n)
    fun(n-1)
    print(n)

fun(10)
```

```python
def fun(n):
  if n == 0:
    return

  print(n)              #Tail recursion
  fun(n-1)              # T.c = O(n)
                        # s.c = O(n)

fun(10)
```
